```python
def calc_adj_abundance (abundance,concentration,tag):
    avogadro_no = 6.022*1e23  
    data['mass'] = data['MW [kDa]'] * abundance * 1e9/ avogadro_no
    total_mass = data['mass'].sum()
    vol_sample = total_mass / concentration
    data['copy_no_per_volume at ' + tag] = abundance / vol_sample
```

```python
def Find_fc_vals(data, col, ref_col, df):
    fc_vals = np.zeros(data.shape[0])
    for row in range(data.shape[0]):
        if data[ref_col].iloc[row] > 0:  
            fc_vals[row] = float(data[col].iloc[row] / data[ref_col].iloc[row])
        else:
            fc_vals[row] = np.nan
    
    tag = col.split('at')[1].strip()
    df['rel_fc_' + tag] = fc_vals
```

```python
def melting_curve(pH, k, xo):
    pH = np.array(pH, dtype=float)
    return 1 / (1 + np.exp((k*xo-k*pH)))
```

```python
def grid_search_with_sse_grid(pH, fold_changes, melting_curve,k_values, xo_values, downweight_thresh=None, downweight_factor=None):
    """
    Grid search over k and xo values to compute SSE grid and best fit.

    Parameters
    ----------
    pH : array_like
        Independent variable (x-axis, e.g., pH values).
    fold_changes : array_like
        Observed data (y-axis).
    melting_curve : callable(pH, k, xo) -> array_like
        Model function to compute predicted values.
    k_values : array_like
        Candidate k values.
    xo_values : array_like
        Candidate xo values.

    Returns
    -------
    dict with keys:
        - best_k : float, k with lowest SSE
        - best_xo : float, xo with lowest SSE
        - best_sse : float, minimum SSE found
        - sse_grid : 2D np.array of SSEs, shape (len(k_values), len(xo_values))
    """
    pH = np.asarray(pH)
    y = np.asarray(fold_changes)

    sse_grid = np.full((len(k_values), len(xo_values)), np.nan)
    best_sse = np.inf
    best_k, best_xo = None, None

     #Compute weights
    weights = np.ones_like(y)
    if downweight_thresh is not None:
        weights[y > downweight_thresh] = downweight_factor 


    for k_i, k in enumerate(k_values):
        for xo_j, xo in enumerate(xo_values):
            pred = melting_curve(pH, k, xo)
            
            residuals = y - pred
            # Weighted SSE
            sse = np.sum(weights * residuals**2)
            sse_grid[k_i, xo_j] = sse

            if sse < best_sse:
                best_sse = sse
                best_k, best_xo = k, xo

    return ( best_k, best_xo, best_sse, sse_grid)
```

```python
def profile_ci_from_sse_grid(k_vals, xo_vals, sse_grid, best_k, best_xo, clean_pH, clean_y, model_func, alpha=0.05):
    """
    k_vals: 1D array of k grid
    xo_vals: 1D array of xo grid
    sse_grid: 2D array shape (len(k_vals), len(xo_vals)) with SSE values
    best_k, best_xo: best-fit parameters (should correspond near the min of sse_grid)
    clean_pH, clean_y: data to compute sigma2
    model_func: callable(x, k, xo)
    Returns dict with 1D CIs for k and xo from grid projection.
    """
    # compute sigma2 from best SSE (use the actual min in sse_grid)
    idx_min = np.unravel_index(np.nanargmin(sse_grid), sse_grid.shape)
    best_sse_grid = sse_grid[idx_min]
    N = len(clean_y)
    p = 2
    sigma2 = best_sse_grid / max(1, N - p)
    delta = sigma2 * chi2.ppf(1 - alpha, df=1)  # threshold

    thresh = best_sse_grid + delta

    # For k: for each k (row), find minimum SSE over xo (i.e., profile)
    profile_xo = np.nanmin(sse_grid, axis=0)  # length len(xo_vals)

    # find where profile <= thresh and take interval
    def extract_interval(vals, profile):
        mask = profile <= thresh
        if not np.any(mask):
            # no crossing found; return NaNs
            return (np.nan, np.nan)
        # Use interpolation to estimate crossings at boundaries for more precision
        # Find contiguous region around the best value index
        best_idx = np.argmin(profile)
        # search left from best_idx to first index where profile > thresh
        left_idx = best_idx
        while left_idx > 0 and profile[left_idx] <= thresh:
            left_idx -= 1
        right_idx = best_idx
        while right_idx < len(profile)-1 and profile[right_idx] <= thresh:
            right_idx += 1

        # build local interpolation on [left_idx, right_idx]
        xi = vals[left_idx:right_idx+1]
        yi = profile[left_idx:right_idx+1]
        # ensure monotonic x for interp; if too small region, fallback to grid endpoints
        try:
            # interpolate crossing on left side
            if left_idx == 0 and yi[0] <= thresh:
                left_val = xi[0]
            else:
                f_left = interp1d(yi[:2], xi[:2], bounds_error=False, fill_value=(xi[0], xi[1]))
                # We want x where y == thresh -> invert: need interp of y->x
                # safer to use interpolation over (xi, yi) and invert with small dense grid
                dense_x = np.linspace(xi[0], xi[-1], 200)
                dense_y = np.interp(dense_x, xi, yi)
                left_cross = dense_x[np.where(dense_y <= thresh)[0][0]]
                left_val = left_cross
        except Exception:
            left_val = vals[left_idx]

        try:
            dense_x = np.linspace(xi[0], xi[-1], 200)
            dense_y = np.interp(dense_x, xi, yi)
            right_cross = dense_x[np.where(dense_y <= thresh)[0][-1]]
            right_val = right_cross
        except Exception:
            right_val = vals[right_idx]

        return (left_val, right_val)


    xo_ci = extract_interval(xo_vals, profile_xo)

    return xo_ci
```

```python
df = df_fc[['Gene Symbol','rel_fc_2.5', 'rel_fc_2.8', 'rel_fc_3.5', 'rel_fc_3.8',
       'rel_fc_4.3', 'rel_fc_5.2', 'rel_fc_5.6', 'rel_fc_5.8', 'rel_fc_6.0',
       'rel_fc_6.8']]
column_names =['protein',2.5,2.8,3.5, 3.8, 4.3, 5.2,5.4, 5.8, 6.0, 6.8]
df.columns = column_names
protein_data = df.drop(columns='protein')
```

```python
# Define ranges of k and xo to test
k_values = np.linspace(1, 20, 200)   
xo_values = np.linspace(1, 6.8, 200) 

pH = np.array([2.5,2.8, 3.5, 3.8, 4.3, 5.2,5.4, 5.8, 6.0, 6.8])
```

```python
results = []

WINDOW_SIZES = [4, 6, 8, 10]

for index, row in protein_data.iterrows():

    fold_changes = row.drop(labels=['protein'], errors='ignore').values.astype(float)

    # Clean NaNs
    mask = ~np.isnan(fold_changes)
    clean_pH = pH[mask]
    clean_fc = fold_changes[mask]

    # Storage for all 4 window sizes
    xo_vals = {w: np.nan for w in WINDOW_SIZES}
    sse_vals = {w: np.nan for w in WINDOW_SIZES}
    ci_lows  = {w: np.nan for w in WINDOW_SIZES}
    ci_highs = {w: np.nan for w in WINDOW_SIZES}


    for w in WINDOW_SIZES: 

      

        # Window
        pH_window = clean_pH[-w:]
        fc_window = clean_fc[-w:]

        # if len(pH_window) < 4:
        #     continue

        try:
            best_k, best_xo, best_sse, sse_grid = grid_search_with_sse_grid(
                pH_window,
                fc_window,
                melting_curve,
                k_values,
                xo_values,downweight_factor=0, downweight_thresh=2.0
            )

            
            ci = profile_ci_from_sse_grid(
                k_values,
                xo_values,
                sse_grid,
                best_k,
                best_xo,
                pH_window,
                fc_window,
                melting_curve,
                alpha=0.05
            )

            # Store results for this window size
            xo_vals[w] = best_xo
            sse_vals[w] = best_sse
            ci_lows[w] = ci[0]
            ci_highs[w] = ci[1]

        except Exception as e:
            print(f"Fit failed (w={w}): {e}")
            continue

    # Save all 4 window-size results as a single row
    results.append((
        xo_vals[4], sse_vals[4], ci_lows[4], ci_highs[4],
        xo_vals[6], sse_vals[6], ci_lows[6], ci_highs[6],
        xo_vals[8], sse_vals[8], ci_lows[8], ci_highs[8],
        xo_vals[10], sse_vals[10], ci_lows[10], ci_highs[10]
    ))
```

```python
results_df = pd.DataFrame(results, columns=['pH501', 'sse1', 'upper_ci1','lower_ci1',
                                            'pH502', 'sse2', 'upper_ci2','lower_ci2',
                                            'pH503', 'sse3', 'upper_ci3','lower_ci3',
                                            'pH504', 'sse4', 'upper_ci4','lower_ci4'])
protein_data_with_results = pd.concat([protein_data.reset_index(drop=True), results_df], axis=1)

protein_data_with_results['# unique peptides']=df_fc['# unique peptides']
protein_data_with_results['# peptides']=df_fc['# peptides']
protein_data_with_results['Description']=df_fc['Description']
protein_data_with_results['Gene Symbol']=df_fc['Gene Symbol']
protein_data_with_results['# PSMs']=df_fc['# PSMs']
protein_data_with_results['Score Sequest HT: Sequest HT']= df_fc['Score Sequest HT: Sequest HT']
protein_data_with_results['calc. pI']= df_fc['calc. pI']
protein_data_with_results['MW [KDa]']= df_fc['MW [KDa]']
protein_data_with_results = protein_data_with_results.dropna(subset=['Gene Symbol'])
fold_cols = [2.5,2.8, 3.5, 3.8, 4.3, 5.2,5.4, 5.8, 6.0, 6.8]

protein_data_with_results = protein_data_with_results.dropna(subset=fold_cols, how='all')

for i in range(1, 5):
    protein_data_with_results[f"ci_width{i}"] = \
        protein_data_with_results[f"lower_ci{i}"] - protein_data_with_results[f"upper_ci{i}"]
best_pH50 = []
best_ci_low = []
best_ci_high = []
best_window = []

for _, row in protein_data_with_results.iterrows():
    ci_widths = []
    
    # collect window results
    for i in range(1, 5):
        ci_w = row[f"ci_width{i}"]
        if not pd.isna(ci_w):  # only valid fits
            ci_widths.append((ci_w, i))

    if len(ci_widths) == 0:
        best_pH50.append(np.nan)
        best_ci_low.append(np.nan)
        best_ci_high.append(np.nan)
        best_window.append(np.nan)
        continue

    # choose window with minimal CI width
    _, best_i = min(ci_widths, key=lambda x: x[0])

    best_pH50.append(row[f"pH50{best_i}"])
    best_ci_low.append(row[f"lower_ci{best_i}"])
    best_ci_high.append(row[f"upper_ci{best_i}"])
    best_window.append(best_i)

protein_data_with_results["best_pH50"] = best_pH50
protein_data_with_results["best_ci_low"] = best_ci_low
protein_data_with_results["best_ci_high"] = best_ci_high
protein_data_with_results["best_window_index"] = best_window
```
